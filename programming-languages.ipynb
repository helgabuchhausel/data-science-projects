```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
import squarify 
import os
```

```python
for dirname, _, filenames in os.walk('./data'):
    for filename in filenames:
        print(os.path.join(dirname, filename))

```

```python
df = pd.read_csv('data/programming_languages.csv')
df
```

```python
sorted_df = df.sort_values(by='popularity', ascending=False)
sorted_df.plot(kind='bar', x='language', y='popularity', legend=False)
figsize=(10, 6)
plt.title('Programming Language Popularity')
plt.xlabel('Programming Language')
plt.ylabel('Popularity')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
top_languages = sorted_df.head(10).sort_values(by='popularity', ascending=True)
plt.figure(figsize=(10, 6))
plt.pie(top_languages['popularity'], labels=top_languages['language'], autopct='%1.1f%%', startangle=140)
plt.title('Programming Languages by Popularity')
```

```python
df_sorted = df.sort_values('popularity', ascending=False)

plt.figure(figsize=(12, 6))

squarify.plot(sizes=df_sorted['popularity'], label=df_sorted['language'], alpha=0.7)

plt.title('Programming Language Popularity (Treemap)')
plt.axis('off') # Hides the axes
plt.show()
```

```python
paradigm_popularity = df.groupby('paradigm')['popularity'].sum().sort_values(ascending=False)
print(paradigm_popularity)

plt.figure(figsize=(10, 6))
sns.barplot(x=paradigm_popularity.index, y=paradigm_popularity
.values)
plt.title('Popularity by Programming Paradigm')
plt.xlabel('Programming Paradigm')
plt.ylabel('Total Popularity')
plt.xticks(rotation=45)
```
